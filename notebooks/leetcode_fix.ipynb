````python
def remove_python_programmer(row):
    remove='You are an expert Python programmer. You will be given a question (problem specification) and will generate a correct Python program that matches the specification and passes all tests.\n\n### Question:\n'
    
    if remove in row['instruction']:
        row['instruction'] = row['instruction'].replace(remove,"")
        
    return row
````

````python
def remove_formate(row):
    row["instruction"] = row["instruction"].str.replace(
        r"### Format:.*$",
        "",
        regex=True
    )
    return row
````
